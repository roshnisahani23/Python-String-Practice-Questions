Q2.Write a Program To Check Whether a String is Palindrome.

```python
s = "nitin"

if s == s[::-1]:
    print(" String is Palindrome")
else:
    print(" String is Not Palindrome")

```

Q3. Write A Program To Count Vowels in Given String.

```python
s = "programming"

count = 0

for i in s:
    if i in "aeiou":
        count = count + 1

print(count)
```

Q8.write A Program to Count Space in a String.

```python
s = "Python is very easy"

count = 0

for i in s:
    if i == " ":
        count = count + 1

print(count)
```
